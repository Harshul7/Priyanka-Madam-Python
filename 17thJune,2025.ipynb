#### Class & Objects 

```python
class fruit:
    def __init__(self,name,colour,taste="Sweet"):
        self.name=name
        self.colour=colour
        self.taste= taste
    
    def getName(self):
        return self.name
    
    def getTaste(self):
        return self.taste
    
    def setName(self,name):
        self.name= name

    def setTaste(self):
        self.taste="Bitter"

fruit1=fruit("Apple","Red")
fruit2=fruit("Mango","Yellow")
fruit3=fruit("Orange","Orange")

print(fruit1)
print(fruit2)

print("fruit1.taste=",fruit1.taste)
fruit1.setTaste()
print("fruit1.taste=",fruit1.getTaste())

```
