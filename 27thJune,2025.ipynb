#### OOPS Concepts

```python
class shape:
    def __init__(self,name,edgeCount):
        self.name=name
        self.edgeCount=edgeCount
    
    def getName(self):
        return self.name
    
    def getEdgeCount(self):
        return self.edgeCount
    
    def setName(self,name):
        self.name=name
    
    def setEdgeCount(self,edgeCnt):
        self.edgeCount=edgeCnt

    def __str__(self):
        return f"{self.name} {self.edgeCount}"
    
shape1=shape("Circle","infinite")
shape2=shape("triangle",3)

print(shape1)
print(shape2)
print("shape1.name=",shape1.getName())
print(shape1.name)
```

```python
class Rectangle:
    def __init__(self,name,length,breadth):
        self.name=name
        self.length=length
        self.breadth=breadth

    def getArea(self):
        return self.length*self.breadth
    
    def getPerimeter(self):
        return 2*(self.length+self.breadth)
    
    def setName(self,nameOfRectangle):
        self.name=nameOfRectangle

    def __str__(self):
        return f"{self.name} Area={self.getArea()}"
    
Rectangle1=Rectangle("Rectangle1",10,5)
print(Rectangle1)
    
```

#### Inheritance 
- Rectangle is Parent class
- Square is Child class
- getDiagonal(self): can not be invoked from Rectangle class
- - Child class' methods and variables can not be invoked from Parent class.
- - Parent class' methods and variables can be invoked from Child class.

```python

class Square(Rectangle):
    def _init_(self,name,edge):
        self.super(name,edge,edge)
    
    def getDiagonal(self):
        return 2**0.5 * (self.length)

Square1=Square("Square1",10,10)

print(Square1.getDiagonal())

print(Square1)
    
```

- Create child classes circle,traingle in class shape

```python
import math 
class Circle(shape):
    def __init__(self,name,edgeCount,radius):
        super().__init__(name,edgeCount)
        #self.super(name,100)
        self.radius = radius
        

    def getName(self):
        return super().getName()
    
    def getCircumference(self):
        return 2*math.pi*self.radius
    
Circle1=Circle("Circle1","Infinite",1)
print(Circle1)

print(Circle1.getCircumference())

print(Circle1.getEdgeCount())
    
    
```
