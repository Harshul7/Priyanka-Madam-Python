#### Multi-level Inheritance


```python
class LivingBeings:
    def __init__(self,name,age,sex):
        self.name=name
        self.age=age
        self.sex=sex
    
    def __str__(self):
        return f"{self.name} {self.age} {self.sex}"
    
    def getName(self):
        return self.name
    
    def setAge(self,newAge):
        self.age=newAge

LivingBeings1=LivingBeings("Sapiens","Infinite","Male")

class HumanBeings(LivingBeings):
    def __init__(self,name,age,sex):
        super().__init__(name,age,sex)
        
class Boys(HumanBeings):
    def __init__(self,name,age):
        super().__init__(name,age,"Male")
        #self.super(name,age,"Male")

Boys1 = Boys("Siddharth",21)
print(Boys1.sex)

```

#### HW
- Implement Multiple Inheritance .
- Single Child and 2 Parents 

```python
class Mother:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    
    def __str__(self):
        return f"Mother {self.name} is of {self.age}."
    
    
class Father:
    def __init__(self,name,age):
        self.name=name
        self.age=age

    def __str__(self):
        return f"Father {self.name} is of {self.age}"
    
    def getFatherName(self):
        return self.name
    
class Kid(Mother,Father):
    def __init__(self,name,age):
        super().__init__(name,age)

Kid1=Kid("Ram",24)

print(Kid1.getFatherName())
print(Kid1)
```

- Multiple Inheritance does not exist in java.
- - Reason: A function,when exists in both the parents but not in child class, 'which function to implement ?' out of the two creates a confusion. In python, first parent's function will be implemented whereas in java, to avoid this confusion, multiple inheritance is not supported.

#### Hybrid Inheritance
- A combination of Multilevel and Multiple Inheritance


```python
class A:
    def __init__(self,name,place):
        self.name=name
        self.place=place

    def __str__(self):
        return f"A={self.name} and place of A ={self.place}"
    
class B(A):
    def __init__(self,name,place):
        super().__init__(name,place)

class C(A):
    def _init_(self,name,animal):
        super().__init__(name)
        self.animal=animal

class D(B,C):
    def __init__(self,name,place,animal,thing):
        super().__init__(name,place)
        self.animal=animal
        self.thing=thing

D1=D("Singh","Gir","Lion","Claws")
B1=B("Lion","Forest")


print(D1)
print(B1)

```
